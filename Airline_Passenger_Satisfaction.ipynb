## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

#### Chart - 1

````python
# Chart 1: Count Plot of Passenger Satisfaction

# Set figure size
plt.figure(figsize=(8,5))

# Create count plot
sns.countplot(
    data=df,
    x='satisfaction',
    palette='Set2',
    edgecolor='black'
)

# Add title and labels
plt.title('Distribution of Passenger Satisfaction', fontsize=14, fontweight='bold')
plt.xlabel('Satisfaction')
plt.ylabel('Number of Passengers')

# Display count values on bars
ax = plt.gca()
for container in ax.containers:
    ax.bar_label(container, fmt='%d', padding=3)

# Show plot
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A Count Plot is the most suitable visualization for the categorical target variable (satisfaction) because it clearly shows the frequency of each category (Satisfied and Neutral/Dissatisfied).

##### 2. What is/are the insight(s) found from the chart?

It shows the overall distribution of passenger satisfaction.
We can easily identify which satisfaction category has more passengers.
It also helps detect whether the dataset is balanced or imbalanced.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Understanding the distribution of customer satisfaction helps airlines evaluate overall service quality. If dissatisfied passengers are more frequent, the airline should focus on improving customer experience, whereas a higher number of satisfied passengers indicates effective service strategies that should be maintained.

#### Chart - 7

````python
# Chart 7: Customer Type vs Satisfaction

customer_sat = pd.crosstab(df['Customer Type'], df['satisfaction'])

customer_sat.plot(
    kind='bar',
    stacked=True,
    figsize=(8,5),
    colormap='Set2',
    edgecolor='black'
)

plt.title('Customer Type vs Passenger Satisfaction')
plt.xlabel('Customer Type')
plt.ylabel('Number of Passengers')
plt.legend(title='Satisfaction')

plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A Stacked Bar Chart effectively compares passenger satisfaction across different customer types while showing the composition of each category in a single visualization.

##### 2. What is/are the insight(s) found from the chart?

Compares satisfaction levels of Loyal Customers and Disloyal Customers.
Identifies which customer type contributes more satisfied passengers.
Highlights differences in customer loyalty and satisfaction.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

This analysis helps airlines evaluate customer loyalty programs and improve services for less satisfied customer groups, leading to higher customer retention and long-term profitability.

#### Chart - 13

````python
# Chart 13: Bubble Plot

plt.figure(figsize=(10,6))

plt.scatter(
    df['Flight Distance'],
    df['Departure Delay in Minutes'],
    s=df['Age']*2,                      # Bubble size
    c=df['Arrival Delay in Minutes'],   # Bubble color
    cmap='viridis',
    alpha=0.6
)

plt.colorbar(label='Arrival Delay (Minutes)')

plt.title('Flight Distance vs Departure Delay')
plt.xlabel('Flight Distance')
plt.ylabel('Departure Delay (Minutes)')

plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A Bubble Plot is suitable for multivariate analysis because it visualizes four variables simultaneously. It helps identify relationships, trends, and patterns among flight distance, departure delay, passenger age, and arrival delay in a single graph.

##### 2. What is/are the insight(s) found from the chart?

Shows the relationship between flight distance and departure delay.
Bubble size represents passenger age, while color indicates arrival delay.
Helps identify clusters and flights with unusually high delays.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

This analysis enables airlines to understand how multiple factors interact and affect operations. It supports better route planning, delay management, and targeted service improvements, ultimately enhancing operational efficiency and passenger satisfaction.

#### Chart - 14 - Correlation Heatmap

````python
# Chart 14: Correlation Heatmap

plt.figure(figsize=(14,10))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Features')

plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A Correlation Heatmap is ideal for multivariate analysis because it displays the correlation among multiple numerical variables in a single visualization. It helps identify strong positive and negative relationships.

##### 2. What is/are the insight(s) found from the chart?

Shows the strength and direction of relationships between numerical features.
Identifies highly correlated variables that may influence passenger satisfaction.
Helps detect features with weak or no correlation.
